```python
#similarity between the phrases
import numpy as np
from typing import List
from rich.console import Console
from rich.table import Table

#returns [-1, +1]

def cosine_similarity(a: List[float], b: List[float]) -> np.float64:
  npa1 = np.array(a)
  npa2 = np.array(b)
  #calculate the cosine of two arrays
  return np.dot(npa1, npa2) / (np.linalg.norm(npa1) * np.linalg.norm(npa2))
```

WHAT ARE EMBEDDINGS

A way to represent meaning with numbers
Text turns into a list of numbers (vector)
Similar ideas -> similar numbers

##Vectors

$$
\vec(a) = (a_1, a_2, \dots, a_n)  
$$

## Dot product - скаларно произведение на два вектора

### Algebraic representation
$$
\vec{a} \cdot \vec{b} = \sum_{i=1}^n{a_i b_i}
$$

### Geometric representation

$$
\vec{a} \cdot \vec{b} = \|\vec{a}\| * \|\vec{b}\| * \cos{\theta}
$$

```python
phrases = ["cat is on the mat", "it is a beuatiful day", "it is sunny today", "dogs love hotdogs", "math is hard", "math is hard?", "math is soo easy"]
```

```python
create_embeddings_response = voyageai_client.embed(
    phrases,
    model="voyage-law-2",
    input_type="document")
```

```python
embeddings = [el for el in create_embeddings_response.embeddings]
```

```python
console = Console()
table = Table("Phrase 1", "Phrase 2", "Similarity")

for i in range(len(phrases)):
    for j in range(i+1, len(phrases)):
        similarity = cosine_similarity(embeddings[i], embeddings[j])  # ← direct indexing, no .embeddings
        table.add_row(phrases[i], phrases[j], str(round(similarity, 4)))

console.print(table)
```
